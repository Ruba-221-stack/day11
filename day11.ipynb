# Task 3: التنفيذ بقرارات مبررة وتوليد الملف النظيف

# 1. الأعمدة التصنيفية القليلة النقص (Gender, Married, Self_Employed): ملء بـ mode
for col in ['Gender', 'Married', 'Self_Employed']:
    df[col].fillna(df[col].mode()[0], inplace=True)

# 2. عمود Dependents: إزالة '+' والتحويل لرقم، ثم ملء النقص بـ mode
df['Dependents'] = df['Dependents'].str.replace('+', '', regex=False)
df['Dependents'].fillna(df['Dependents'].mode()[0], inplace=True)
df['Dependents'] = pd.to_numeric(df['Dependents'])

# 3. عمود LoanAmount: ملء بالوسيط (median)
df['LoanAmount'].fillna(df['LoanAmount'].median(), inplace=True)

# 4. عمود Loan_Amount_Term: ملء بالقيمة الأكثر تكراراً (mode)
df['Loan_Amount_Term'].fillna(df['Loan_Amount_Term'].mode()[0], inplace=True)

# 5. عمود Credit_History: حساب نسبة الموافقة قبل/بعد ثم ملء بـ (1)
print("نسبة الموافقة حسب Credit_History قبل الملء:")
print(df.groupby('Credit_History')['Loan_Status'].value_counts(normalize=True))
df['Credit_History'].fillna(1, inplace=True)

# 6. التعامل مع Outliers (إبقاء القيم المتطرفة في ApplicantIncome و LoanAmount وتوثيقها بناءً على القرار)
# تم الاكتفاء بالدراسة والإبقاء عليها لعدم فقدان البيانات الحقيقية.

# 7. تحويل الـ Target (Loan_Status) إلى أرقام (Y=1, N=0)
df['Loan_Status'] = df['Loan_Status'].map({'Y': 1, 'N': 0})

# 8. التأكد من أن عدد القيم الناقصة أصبح صفراً لكل الأعمدة
print("\n--- عدد القيم الناقصة بعد التنظيف ---")
print(df.isnull().sum())

# 9. حفظ الملف النظيف
df.to_csv("loan_clean.csv", index=False)
print("\nتم حفظ الملف بنجاح باسم: loan_clean.csv")

```python
from IPython.display import Markdown, display

log = []  # (القرار، السبب بالأرقام، المتأثر)

# 1. الأعمدة التصنيفية القليلة النقص: mode
for col in ["Gender", "Married", "Self_Employed"]:
    m = df[col].mode()[0]
    n = df[col].isna().sum()
    print(f"{col}: ناقص = {n} | الأكثر تكرارًا = {m}")
    df[col] = df[col].fillna(m)
    log.append((f"{col}: الملء بـ {m}", f"الأكثر تكرارًا والنقص قليل", f"{n} خلية"))

# 2. Dependents: ملء ثم 3+ إلى 3 ثم رقم
n = df["Dependents"].isna().sum()
n3 = (df["Dependents"] == "3+").sum()
m = df["Dependents"].mode()[0]
print(f"Dependents: ناقص = {n} | الأكثر تكرارًا = {m} | عدد 3+ = {n3}")
df["Dependents"] = df["Dependents"].fillna(m).replace("3+", "3").astype(int)
log.append(("Dependents: mode ثم 3+ إلى 3 ثم int",
            f"3+ نص يمنع الحساب، والأكثر تكرارًا {m}", f"{n} ناقصة + {n3} قيمة 3+"))

# 3. LoanAmount: طباعة المتوسط والوسيط أولًا
mean_v, med_v = df["LoanAmount"].mean(), df["LoanAmount"].median()
n = df["LoanAmount"].isna().sum()
print(f"LoanAmount: المتوسط = {mean_v:.1f} | الوسيط = {med_v} | ناقص = {n}")
df["LoanAmount"] = df["LoanAmount"].fillna(med_v)
log.append(("LoanAmount: الملء بالوسيط",
            f"المتوسط {mean_v:.1f} أكبر من الوسيط {med_v} (توزيع مائل)", f"{n} خلية"))

# 4. Loan_Amount_Term
t = df["Loan_Amount_Term"].mode()[0]
n = df["Loan_Amount_Term"].isna().sum()
print(f"Loan_Amount_Term: الأكثر تكرارًا = {t:.0f} شهر = {t/12:.0f} سنة | ناقص = {n}")
df["Loan_Amount_Term"] = df["Loan_Amount_Term"].fillna(t)
log.append((f"Loan_Amount_Term: الملء بـ {t:.0f}", f"الأكثر تكرارًا = {t/12:.0f} سنة", f"{n} خلية"))

# 5. Credit_History: فحص الناقص أولًا (fillna(-1) مؤقت)
temp = df.copy()
temp["Credit_History"] = temp["Credit_History"].fillna(-1)
temp["approved"] = temp["Loan_Status"].map({"Y": 1, "N": 0})
rates = temp.groupby("Credit_History")["approved"].agg(["mean", "count"]).round(3)
print(rates)
r_miss, r0, r1 = rates.loc[-1, "mean"], rates.loc[0.0, "mean"], rates.loc[1.0, "mean"]
n = df["Credit_History"].isna().sum()
df["Credit_History"] = df["Credit_History"].fillna(1.0)   # عدّلها لو نتيجتك تقول غير ذلك
log.append(("Credit_History: الملء بـ 1",
            f"موافقة الناقص {r_miss:.0%} أقرب لموافقة الواحد ({r1:.0%}) من الصفر ({r0:.0%})",
            f"{n} خلية"))

# 6. Outliers بطريقة IQR (إبقاء)
for col in ["ApplicantIncome", "CoapplicantIncome", "LoanAmount"]:
    q1, q3 = df[col].quantile(0.25), df[col].quantile(0.75)
    iqr = q3 - q1
    low, high = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    n = ((df[col] < low) | (df[col] > high)).sum()
    print(f"{col}: [{low:.0f}, {high:.0f}] | Outliers = {n} ({n/len(df)*100:.1f}%)")
    log.append((f"{col}: إبقاء الـ Outliers",
                f"عددها {n} ({n/len(df)*100:.1f}%)، دخول ومبالغ حقيقية وليست أخطاء", "0 صف محذوف"))

# 7. تحويل Target
df["Loan_Status"] = df["Loan_Status"].map({"Y": 1, "N": 0})
log.append(("Loan_Status: Y=1 و N=0", "الموديل يحتاج رقمًا", f"{len(df)} صف"))

# 8. التأكد والحفظ
print("\nمجموع القيم الناقصة:", df.isna().sum().sum())
print(df.shape)
df.to_csv("loan_clean.csv", index=False)

# 9. Decision Log جاهز للنسخ
table = "### Decision Log\n\n| القرار | السبب بالأرقام | المتأثر |\n|---|---|---|\n"
table += "\n".join(f"| {a} | {b} | {c} |" for a, b, c in log)
display(Markdown(table))
```

الخلاصة:

ساهمت عملية تنظيف البيانات ومعالجتها في تحسين جودة سجلات الحفريات، وجعلها أكثر دقة وتنظيماً. كما ساعد التحليل على معرفة متوسط أطوال الأنواع وتحديد الأنواع والدول الأكثر تكراراً، مما يسهل الاستفادة من البيانات في الدراسات المستقبلية

```python
import pandas as pd

fossils = {
    "fossil_id": ["F01", "F02", "F03", "F04", "F05", "F02", "F06", "F07", "F08", "F09"],
    "species": ["T-Rex", "Triceratops", "t-rex", "Velociraptor", "Stegosaurus", "Triceratops",
                "Brachiosaurus", "T-Rex", "Velociraptor", "Stegosaurus"],
    "age_million_years": [68, 67, 66, 75, -150, 67, 150, 68, None, 155],
    "length": ["12 m", "9 m", "40 ft", "2 m", "9 m", "9 m", "26 m", "12.3 m", "6.6 ft", "9 m"],
    "found_in": ["USA", "USA", "Canada", "Mongolia", "USA", "USA", "Tanzania", "USA", "Mongolia", "Portugal"],
}
df_f = pd.DataFrame(fossils)

# 1. حذف المكرر وتوحيد أسماء الأنواع
print("الصفوف المكررة:", df_f.duplicated().sum())
df_f = df_f.drop_duplicates().reset_index(drop=True)
df_f["species"] = df_f["species"].str.strip().str.title()
print(df_f["species"].value_counts())

# 2. العمر السالب
print("\nأعمار سالبة:", (df_f["age_million_years"] < 0).sum())
df_f["age_million_years"] = df_f["age_million_years"].abs()

# 3. ملء العمر الناقص بوسيط نفس النوع
print("أعمار ناقصة:", df_f["age_million_years"].isna().sum())
df_f["age_million_years"] = df_f["age_million_years"].fillna(
    df_f.groupby("species")["age_million_years"].transform("median")
)

# 4. فصل الرقم عن الوحدة والتحويل إلى متر
parts = df_f["length"].str.split(" ", expand=True)
df_f["length_value"] = parts[0].astype(float)
df_f["unit"] = parts[1]
df_f["m_length"] = df_f["length_value"].where(df_f["unit"] == "m", df_f["length_value"] * 0.3048).round(2)
display(df_f)

# 5. التحليل
print("\nمتوسط الطول لكل نوع (متر):")
print(df_f.groupby("species")["m_length"].mean().round(2).sort_values(ascending=False))

oldest = df_f.loc[df_f["age_million_years"].idxmax()]
print("\nأقدم نوع:", oldest["species"], "-", oldest["age_million_years"], "مليون سنة")

country_counts = df_f["found_in"].value_counts()
print("أكثر دولة:", country_counts.idxmax(), "-", country_counts.max(), "حفريات")
```

**لو كنت مسؤولًا عن نظام الإدخال:**
